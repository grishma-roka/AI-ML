```python
import re

def lower_case(text):
  return text.lower()

def remove_urls(text):
  return re.sub(r'https?://\S+|www\.\S+', '', text)

def remove_unwanted_char(text):
  # FIXED: Added the closing bracket "]" at the end of the final unicode range
  emoji_pattern = re.compile("["
                           u"\U0001F600-\U0001F64F"  # emoticons
                           u"\U0001F300-\U0001F5FF"  # symbols & pictographs
                           u"\U0001F680-\U0001F6FF"  # transport & map symbols
                           u"\U0001F1E0-\U0001F1FF"  # flags (iOS)
                           u"\U00002702-\U000027B0"
                           u"\U000024C2-\U0001F251"
                           "]") # <--- Added this line

  text = emoji_pattern.sub(r'', text)
  # remove # and @
  text = re.sub(r'[#@]\w+', '', text)
  text = re.sub(r'[^\w\s]', '', text)
  text = re.sub(r'\s+',' ',text)

  return text
```

```python
def text_cleaning_pipeline(dataset, rule = "lemmatize"):
  """
  This...
  """
  # Convert the input to small/lower order.
  data = lower_case(text)
  # Remove URLs
  data = remove_urls(data)
  # Remove emojis
  data = remove_unwanted_char(data)
  # Remove all other unwanted characters.

  # Create tokens.
  tokens = tokenize_stop_word(data)
  # Remove stopwords:

  if rule == "lemmatize":
    tokens = lame_stem(tokens, lemm = True)
  elif rule == "stem":
    tokens = lame_stem(tokens, lemm = True)
  else:
    print("Pick between lemmatize or stem")


  return " ".join(tokens)
```
